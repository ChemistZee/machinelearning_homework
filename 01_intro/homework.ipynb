```python
#working on the df_asia dataset
df_asia2 = df_asia[["vehicle_weight", "model_year"]]
```

```python
X = df_asia2.head(7).to_numpy()
```

```python
X
```

```python
#matrix operation
TX = X.T
```

```python
TX
```

```python
XTX = TX @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
matrix = XTX_inv @ TX
matrix
```

```python
w = matrix @ y
w
```
